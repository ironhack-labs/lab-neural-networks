### Observations

- Test loss reached **0.030**, which is below the required target of 0.05 ✓
- Training loss reached **0.000**, meaning the model perfectly memorised
  the training set
- The gap between training loss (0.000) and test loss (0.030) indicates
  mild overfitting — the model still generalises well on unseen data
- All 7 features were necessary — the sine features (sin(X₁) and sin(X₂))
  were critical for capturing the rotational pattern of the spiral
- Running beyond ~1000 epochs showed diminishing returns as training loss
  was already floored at 0.000
